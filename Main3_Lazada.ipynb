```python
# Customize chrome display
chrome_options = Options()
chrome_options.add_argument("--no-sandbox")
#chrome_options.add_argument("--headless") # run tanpa membuka chrome
chrome_options.add_argument("disable-notifications")
chrome_options.add_argument("--disable-infobars")
chrome_options.add_experimental_option("excludeSwitches", ['enable-automation'])

service = Service(path)
options = webdriver.ChromeOptions()


driver = webdriver.Chrome(service=service, options=options)
driver.get(main_link)
```

```python
""" Product Name """
try:
    # Scroll down to load more content if needed
    driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
    import time
    time.sleep(25)  # Wait for content to load
    
    # Get fresh HTML after scrolling
    html = driver.execute_script("return document.documentElement.outerHTML")
    soup = BeautifulSoup(html, "html.parser")
    
    # Use CSS selector to find product names
    css_selector = "div._17mcb div.Bm3ON div.RfADt"
    name_elements = soup.select(css_selector)
    
    # Process the found elements
    if name_elements:
        all_product_names = [element.get_text().strip() for element in name_elements]
        print(f"Found {len(all_product_names)} product names")
        
        # Print a sample of results (first 5 or less)
        for i in range(min(5, len(all_product_names))):
            print(f"Name {i+1}: {all_product_names[i]}")
    else:
        # Try direct class search as fallback
        name_elements = soup.find_all('div', class_='RfADt')
        if name_elements:
            all_product_names = [element.get_text().strip() for element in name_elements]
            print(f"Found {len(all_product_names)} product names with direct search")
            
            # Print a sample of results
            for i in range(min(5, len(all_product_names))):
                print(f"Name {i+1}: {all_product_names[i]}")
        else:
            print("No product names found. The website structure might have changed.")
        
except Exception as e:
    print(f"Error extracting product names: {e}")
```

```python
""" Product Sold """
try:
    # Scroll down to load more content if needed
    driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
    import time
    time.sleep(2)  # Wait for content to load
    
    # Get fresh HTML after scrolling
    html = driver.execute_script("return document.documentElement.outerHTML")
    soup = BeautifulSoup(html, "html.parser")
    
    # First, get all product containers to know how many products there are
    product_containers = soup.select("div._17mcb div.Bm3ON")
    total_products = len(product_containers)
    print(f"Found {total_products} product containers")
    
    # Use CSS selector to find product sold counts
    css_selector = "div._17mcb div.Bm3ON span._1cEkb"
    sold_elements = soup.select(css_selector)
    
    # Process the found elements
    if sold_elements:
        all_sold_counts = [element.get_text().strip() for element in sold_elements]
        print(f"Found {len(all_sold_counts)} products with sold counts")
        
        # Check if some products don't have sold counts
        if len(all_sold_counts) < total_products:
            missing_count = total_products - len(all_sold_counts)
            print(f"Adding '0' as sold count for {missing_count} products without sold counts")
            
            # Create a mapping from product container to sold count
            product_to_sold = {}
            
            # For each product container, try to find its sold count
            for i, container in enumerate(product_containers):
                # Try to find the sold count element within this specific container
                sold_element = container.select_one("span._1cEkb")
                if sold_element:
                    product_to_sold[i] = sold_element.get_text().strip()
                else:
                    product_to_sold[i] = "0"  # Assign "0" if no sold count found
            
            # Create the final list in the correct order
            all_sold_counts = [product_to_sold[i] for i in range(total_products)]
        
        print(f"Final count: {len(all_sold_counts)} product sold counts")
        
        # Print a sample of results (first 5 or less)
        for i in range(min(5, len(all_sold_counts))):
            print(f"Sold {i+1}: {all_sold_counts[i]}")
    else:
        print("No product sold counts found. Assigning '0' to all products.")
        all_sold_counts = ["0"] * total_products
        
        # Print a sample of results
        for i in range(min(5, len(all_sold_counts))):
            print(f"Sold {i+1}: {all_sold_counts[i]}")
        
except Exception as e:
    print(f"Error extracting product sold counts: {e}")
```

```python
""" Product Link """
try:
    # Scroll down to load more content if needed
    driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
    time.sleep(2)  # Wait for content to load
    
    # Get fresh HTML after scrolling
    html = driver.execute_script("return document.documentElement.outerHTML")
    soup = BeautifulSoup(html, "html.parser")
    
    # First, get all product containers to know how many products there are
    product_containers = soup.select("div._17mcb div.Bm3ON")
    total_products = len(product_containers)
    print(f"Found {total_products} product containers")
    
    # Initialize list for product links
    all_product_links = []
    
    # Try multiple class approaches to find product links
    for container in product_containers:
        product_link = None
        
        # Approach 1: Try the first class
        link_element = container.select_one("div._95X4G a")
        if link_element and 'href' in link_element.attrs:
            product_link = link_element['href']
        
        # Approach 2: Try the second class if first didn't work
        if not product_link:
            link_element = container.select_one("div._95X4G a")
            if link_element and 'href' in link_element.attrs:
                product_link = link_element['href']
        
        # Approach 3: Try any a tag as fallback
        if not product_link:
            all_links = container.find_all('a')
            for link in all_links:
                if 'href' in link.attrs and '/p/' in link['href']:  # Tokopedia product links typically contain /p/
                    product_link = link['href']
                    break
        
        # Make sure URLs are absolute
        if product_link and not product_link.startswith('http'):
            if product_link.startswith('//'):
                product_link = 'https:' + product_link
            elif product_link.startswith('/'):
                product_link = 'https://www.tokopedia.com' + product_link
        
        # Add to list (use empty string if no link found)
        all_product_links.append(product_link if product_link else "")
    
    # Count products with links
    valid_links = [link for link in all_product_links if link]
    print(f"Found {len(valid_links)} product links out of {total_products} products")
    
    # Print a sample of results (first 5 or less)
    for i in range(min(5, len(all_product_links))):
        if all_product_links[i]:
            print(f"Link {i+1}: {all_product_links[i]}")
        else:
            print(f"Link {i+1}: No link found")
    
except Exception as e:
    print(f"Error extracting product links: {e}")
```
