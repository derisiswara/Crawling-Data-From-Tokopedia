```python
""" Product Sold """
try:
    # Scroll down to load more content if needed
    driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
    import time
    time.sleep(2)  # Wait for content to load
    
    # Get fresh HTML after scrolling
    html = driver.execute_script("return document.documentElement.outerHTML")
    soup = BeautifulSoup(html, "html.parser")
    
    # First, get all product containers to know how many products there are
    product_containers = soup.select("div.css-jza1fo div.css-5wh65g")
    total_products = len(product_containers)
    print(f"Found {total_products} product containers")
    
    # Use CSS selector to find product sold counts
    css_selector = "div.css-jza1fo div.css-5wh65g div.Lrp\\+JcoWPuzTgMQ41Mkg3w\\=\\="
    sold_elements = soup.select(css_selector)
    
    # Process the found elements
    if sold_elements:
        all_sold_counts = [element.get_text().strip() for element in sold_elements]
        print(f"Found {len(all_sold_counts)} products with sold counts")
        
        # Check if some products don't have sold counts
        if len(all_sold_counts) < total_products:
            missing_count = total_products - len(all_sold_counts)
            print(f"Adding '0' as sold count for {missing_count} products without sold counts")
            
            # Create a mapping from product container to sold count
            product_to_sold = {}
            
            # For each product container, try to find its sold count
            for i, container in enumerate(product_containers):
                # Try to find the sold count element within this specific container
                sold_element = container.select_one("div.Lrp\\+JcoWPuzTgMQ41Mkg3w\\=\\=")
                if sold_element:
                    product_to_sold[i] = sold_element.get_text().strip()
                else:
                    product_to_sold[i] = "0"  # Assign "0" if no sold count found
            
            # Create the final list in the correct order
            all_sold_counts = [product_to_sold[i] for i in range(total_products)]
        
        print(f"Final count: {len(all_sold_counts)} product sold counts")
        
        # Print a sample of results (first 5 or less)
        for i in range(min(5, len(all_sold_counts))):
            print(f"Sold {i+1}: {all_sold_counts[i]}")
    else:
        print("No product sold counts found. Assigning '0' to all products.")
        all_sold_counts = ["0"] * total_products
        
        # Print a sample of results
        for i in range(min(5, len(all_sold_counts))):
            print(f"Sold {i+1}: {all_sold_counts[i]}")
        
except Exception as e:
    print(f"Error extracting product sold counts: {e}")
```

```python
""" Product Location """
try:
    # Scroll down to load more content if needed
    driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
    import time
    time.sleep(2)  # Wait for content to load
    
    # Get fresh HTML after scrolling
    html = driver.execute_script("return document.documentElement.outerHTML")
    soup = BeautifulSoup(html, "html.parser")
    
    # First, get all product containers to know how many products there are
    product_containers = soup.select("div.css-jza1fo div.css-5wh65g")
    total_products = len(product_containers)
    print(f"Found {total_products} product containers")
    
    # Use CSS selector to find product locations using span element instead of div
    css_selector = "div.css-jza1fo div.css-5wh65g span.pC8DMVkBZGW7\\-egObcWMFQ\\=\\="
    location_elements = soup.select(css_selector)
    
    # Process the found elements
    if location_elements:
        # Extract text from all elements
        all_raw_locations = [element.get_text().strip() for element in location_elements]
        print(f"Found {len(all_raw_locations)} total location elements")
        
        # Take every second element (odd indexed elements: 1, 3, 5, etc.)
        # In Python, this is done with slicing [1::2]
        all_locations = all_raw_locations[1::2]
        print(f"Extracted {len(all_locations)} locations (every second element)")
        
        # Make sure we have enough locations for all products
        if len(all_locations) < total_products:
            # Add unknown locations to match product count
            all_locations.extend(["Unknown Location"] * (total_products - len(all_locations)))
        
        print(f"Final count: {len(all_locations)} product locations")
        
        # Print a sample of results (first 5 or less)
        for i in range(min(5, len(all_locations))):
            print(f"Location {i+1}: {all_locations[i]}")
            
        # Also print the first few raw locations to verify the pattern
        print("\nFirst few raw location elements:")
        for i in range(min(6, len(all_raw_locations))):
            print(f"Raw {i}: {all_raw_locations[i]}")
    else:
        print("No product locations found. Assigning 'Unknown Location' to all products.")
        all_locations = ["Unknown Location"] * total_products
        
        # Print a sample of results
        for i in range(min(5, len(all_locations))):
            print(f"Location {i+1}: {all_locations[i]}")
        
except Exception as e:
    print(f"Error extracting product locations: {e}")
```

```python
""" Product Rating """
try:
    # Scroll down to load more content if needed
    driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
    import time
    time.sleep(2)  # Wait for content to load
    
    # Get fresh HTML after scrolling
    html = driver.execute_script("return document.documentElement.outerHTML")
    soup = BeautifulSoup(html, "html.parser")
    
    # First, get all product containers to know how many products there are
    product_containers = soup.select("div.css-jza1fo div.css-5wh65g")
    total_products = len(product_containers)
    print(f"Found {total_products} product containers")
    
    # Use CSS selector to find product ratings
    css_selector = "div.css-jza1fo div.css-5wh65g span._9jWGz3C\\-GX7Myq\\-32zWG9w\\=\\="
    rating_elements = soup.select(css_selector)
    
    # Process the found elements
    if rating_elements:
        all_ratings = [element.get_text().strip() for element in rating_elements]
        print(f"Found {len(all_ratings)} products with rating information")
        
        # Check if some products don't have rating info
        if len(all_ratings) < total_products:
            missing_count = total_products - len(all_ratings)
            print(f"Adding 'No Rating' for {missing_count} products without rating info")
            
            # Create a mapping from product container to rating
            product_to_rating = {}
            
            # For each product container, try to find its rating
            for i, container in enumerate(product_containers):
                # Try to find the rating element within this specific container
                rating_element = container.select_one("span._9jWGz3C\\-GX7Myq\\-32zWG9w\\=\\=")
                if not rating_element:
                    # Try direct class search if CSS selector fails
                    rating_elements_direct = container.find_all('span', class_='_9jWGz3C-GX7Myq-32zWG9w==')
                    if rating_elements_direct:
                        rating_element = rating_elements_direct[0]
                
                if rating_element:
                    product_to_rating[i] = rating_element.get_text().strip()
                else:
                    product_to_rating[i] = "No Rating"  # Assign default if no rating found
            
            # Create the final list in the correct order
            all_ratings = [product_to_rating[i] for i in range(total_products)]
        
        print(f"Final count: {len(all_ratings)} product ratings")
        
        # Print a sample of results (first 5 or less)
        for i in range(min(5, len(all_ratings))):
            print(f"Rating {i+1}: {all_ratings[i]}")
            
        # Try to convert ratings to float for statistical analysis
        numeric_ratings = []
        for rating in all_ratings:
            try:
                # Extract just the numeric part if there's additional text
                if rating != "No Rating":
                    # Some common patterns: "4.5", "4,5", "4.5/5", "4.5 (123)"
                    rating_value = rating.split('/')[0].split('(')[0].strip()
                    rating_value = rating_value.replace(',', '.')  # Handle comma as decimal separator
                    numeric_ratings.append(float(rating_value))
            except ValueError:
                # Skip if can't convert to float
                pass
                
        if numeric_ratings:
            avg_rating = sum(numeric_ratings) / len(numeric_ratings)
            print(f"\nStatistics for {len(numeric_ratings)} numeric ratings:")
            print(f"Average rating: {avg_rating:.2f}")
            print(f"Min rating: {min(numeric_ratings)}")
            print(f"Max rating: {max(numeric_ratings)}")
        
    else:
        print("No product ratings found. Assigning 'No Rating' to all products.")
        all_ratings = ["No Rating"] * total_products
        
        # Print a sample of results
        for i in range(min(5, len(all_ratings))):
            print(f"Rating {i+1}: {all_ratings[i]}")
        
except Exception as e:
    print(f"Error extracting product ratings: {e}")
```

```python
""" Product Link """
try:
    # Scroll down to load more content if needed
    driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
    import time
    time.sleep(2)  # Wait for content to load
    
    # Get fresh HTML after scrolling
    html = driver.execute_script("return document.documentElement.outerHTML")
    soup = BeautifulSoup(html, "html.parser")
    
    # First, get all product containers to know how many products there are
    product_containers = soup.select("div.css-jza1fo div.css-5wh65g")
    total_products = len(product_containers)
    print(f"Found {total_products} product containers")
    
    # Initialize list for product links
    all_product_links = []
    
    # Try multiple class approaches to find product links
    for container in product_containers:
        product_link = None
        
        # Approach 1: Try the first class
        link_element = container.select_one("a.oQ94Awb6LlTiGByQZo8Lyw\\=\\=")
        if link_element and 'href' in link_element.attrs:
            product_link = link_element['href']
        
        # Approach 2: Try the second class if first didn't work
        if not product_link:
            link_element = container.select_one("a.IM26HEnTb\\-krJayD\\-R0OHw\\=\\=")
            if link_element and 'href' in link_element.attrs:
                product_link = link_element['href']
        
        # Approach 3: Try any a tag as fallback
        if not product_link:
            all_links = container.find_all('a')
            for link in all_links:
                if 'href' in link.attrs and '/p/' in link['href']:  # Tokopedia product links typically contain /p/
                    product_link = link['href']
                    break
        
        # Make sure URLs are absolute
        if product_link and not product_link.startswith('http'):
            if product_link.startswith('//'):
                product_link = 'https:' + product_link
            elif product_link.startswith('/'):
                product_link = 'https://www.tokopedia.com' + product_link
        
        # Add to list (use empty string if no link found)
        all_product_links.append(product_link if product_link else "")
    
    # Count products with links
    valid_links = [link for link in all_product_links if link]
    print(f"Found {len(valid_links)} product links out of {total_products} products")
    
    # Print a sample of results (first 5 or less)
    for i in range(min(5, len(all_product_links))):
        if all_product_links[i]:
            print(f"Link {i+1}: {all_product_links[i]}")
        else:
            print(f"Link {i+1}: No link found")
    
except Exception as e:
    print(f"Error extracting product links: {e}")
```
